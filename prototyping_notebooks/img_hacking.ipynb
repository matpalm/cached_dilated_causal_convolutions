```python
W, H = 489, 360
N = 7
B = 2  # buffer between image

collage = Image.new('RGB', (N*(W+B), N*(H+B)), (0,0,0))

for i in range(N):
    for j in range(N):
        fname = f"../foo_{i:02d}_{j:02d}.png"
        img = Image.open(fname).convert('RGB')        
        img = img.crop((84, 63, 573, 423))
        collage.paste(img, (i*(W+B), j*(H+B)))

collage.save('interp_examples.png')
```
